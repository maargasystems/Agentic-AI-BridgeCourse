```python
import os
from dotenv import load_dotenv
load_dotenv(override=True)

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
os.environ["LANGCHAIN_API_KEY"]=os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_PROJECT"]=os.getenv("LANGCHAIN_PROJECT")
os.environ["LANGCHAIN_TRACING_V2"]="true"
```

```python
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

model = init_chat_model(model="groq:openai/gpt-oss-20b")
model
```

```python
from langchain_groq.chat_models import ChatGroq
model = ChatGroq(model="qwen/qwen3.6-27b")
model
```

```python
response = model.invoke("who is saran?")
response
```

```python
response.content
```

```python
for chunk in model.stream("who is saran?"):
    print(chunk.text, end= "|", flush=True)
```

```python
for chunk in model.stream("why the sky is blue?"):
    print(chunk.text, end="|", flush=True)
```

```python
response = model.batch(["what does maarga systems do?", "what does vajra solutions do?", "can you tell a more about the prime gym located in dhandesswaram, chennai"])
for i in response:
    print(i)
```

```python
response = model.batch(["what is the weather in chennai", "when facebook was introduced", "who is the father of india",
                        "what is the national sport of england?"], config={"max_concurrency":3})
response
```
